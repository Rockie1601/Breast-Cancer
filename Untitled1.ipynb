```python
X=df.iloc[:,:-1]
y=df.iloc[:,-1]
X.shape,y.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
X_train.shape,y_train.shape
```

```python
from sklearn.preprocessing import LabelEncoder
l1=LabelEncoder()
l2=LabelEncoder()
l3=LabelEncoder()
l4=LabelEncoder()

X_train['A Stage']=l1.fit_transform(X_train['A Stage'])
X_train['Estrogen Status']=l2.fit_transform(X_train['Estrogen Status'])
X_train['Progesterone Status']=l3.fit_transform(X_train['Progesterone Status'])
y_train=l4.fit_transform(y_train)

X_test['A Stage']=l1.transform(X_test['A Stage'])
X_test['Estrogen Status']=l2.transform(X_test['Estrogen Status'])
X_test['Progesterone Status']=l3.transform(X_test['Progesterone Status'])
y_test=l4.transform(y_test)
```

```python
X_train.info()
```

```python
encoder_col=Colu[:5]
print(encoder_col)
```

```python
column_indices = [X_train.columns.get_loc(name) for name in encoder_col]
print(column_indices)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct=ColumnTransformer(transformers=[('onehotencoding', OneHotEncoder(),column_indices)],remainder='passthrough')
X_train=ct.fit_transform(X_train)
X_test=ct.transform(X_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
```

```python
params={
    'criterion':['gini', 'entropy', 'log_loss'],
    'splitter':['best','random'],
    'max_depth':[5,10,15,20,25,30,40,50],
    'max_features':['sqrt','log2']
}
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(dt,params)
```

```python
import warnings
warnings.filterwarnings('ignore')
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred=grid.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
as_grid=accuracy_score(y_test,y_pred)
print(as_grid)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
pickle.dump(grid,open('model.pkl','wb'))
pickle.dump(ct,open('ct.pkl','wb'))
pickle.dump(l1,open('l1.pkl','wb'))
pickle.dump(l2,open('l2.pkl','wb'))
pickle.dump(l3,open('l3.pkl','wb'))
pickle.dump(l4,open('l4.pkl','wb'))
```
